```python
def relu(x, derivative=False):
        """ReLU activation function with proper derivative handling"""
        x = np.array(x)  # Ensure x is a NumPy array
        if derivative is True:  # Ensure derivative is a boolean, not an array
            return np.where(x > 0, 1.0, 0.0)  
        return np.maximum(0, x)
```
